```python
# ==============================================================================
# NOTEBOOK CELL 2: DATABASE INGESTION & 3D ATTRIBUTED GRAPH CONVERSION
# ==============================================================================
# Objective:
# Retrieve baseline chemical structures (D_chem) and historical validation records
# (D_val) from the external relational storage backend (DrugBank MySQL), generate
# 3D Cartesian coordinates R via distance geometry, and compile attributed spatial
# graphs G = (V, E, X, R) ready for molecular geometry encoding.
#
# Patent Module Mapping:
# - Input Preparation for System 100
# - Attributed Spatial Molecular Graph compilation: G = (V, E, X, R) for Module 102
# - Corpus preparation:
#     * Baseline chemical structure corpus (D_chem) for unsupervised reconstruction
#       and structural Fisher Information evaluation in Module 108
#     * Historical validation dataset (D_val) maintained for empirical feedback
#       processing and predictive Fisher Information evaluation in Module 108
#
# Input Data:
# External relational database tables: `drug`, `drug_property`, `interactant`,
# `interactant_polypeptide`, and `polypeptide`.
#
# Processing Steps:
# 1. Establish socket connection to MySQL database via PyMySQL with DictCursor.
# 2. Extract small-molecule chemical structures (D_chem) from `drug` and `drug_property`
#    using exact schema match `kind = 'SMILES'` with length constraint (< 250 chars).
# 3. Extract historical drug-target validation records (D_val) joining on composite key
#    (polypeptide_id, source).
# 4. Apply demonstration sanitization (parent organic fragment selection, canonicalization).
# 5. Generate Cartesian coordinates R in R^(N x 3) using RDKit ETKDGv3 as one possible
#    implementation method, with planar 2D projection fallback.
# 6. Apply optional demonstration coordinate centering: R <- R - R_mean.
# 7. Extract atomic feature matrix X in R^(N x F_v) and edge tensor E with attributes e_ij.
# 8. Instantiate attributed spatial graph instances G = (V, E, X, R) as torch_geometric.data.Data.
# 9. Partition baseline corpus into training and validation subsets for continual learning evaluation.
#
# Mathematical Operation:
# Attributed spatial graph formalization (Patent FIG. 3):
#   G = (V, E, X, R)
# Optional coordinate origin centering (demonstration preprocessing):
#   R_i <- R_i - (1 / N) * sum_{j=1}^N R_j
#
# Expected Output:
# Diagnostic summaries verifying retrieval of baseline chemical structure records (D_chem),
# historical validation records (D_val), and successful tensor instantiation of attributed graphs G.
#
# Saved Files / Results:
# In-memory datasets: `graph_corpus_chem` (D_chem), `train_graphs`, `validation_graphs`,
# `val_records_df` (D_val), and tensor dimensions `NODE_FEAT_DIM` (F_v), `EDGE_FEAT_DIM` (F_e).
#
# Computational Notes:
# - Patent-Disclosed Technical Features:
#     * Representation of chemical entities as attributed spatial graphs G = (V, E, X, R)
#     * Parity-sensitive chiral indicators in X and stereochemical configuration flags in e_ij
#     * Preservation of D_chem and D_val for regularized adaptation in Module 108
# - Non-Claimed Implementation Choices:
#     * DrugBank relational schema (used purely as an external data storage backend)
#     * RDKit and ETKDGv3 conformer algorithm (one possible embodiment for obtaining coordinates R)
#     * PyTorch Geometric Data containers and Morgan fingerprints
#     * Specific coordinate origin centering and atom count thresholds (optional preprocessing)
# ==============================================================================

import os
import sys
import random
import logging
import warnings
import pymysql
from pymysql.cursors import DictCursor

import numpy as np
import pandas as pd
from tqdm.auto import tqdm

import torch
from torch_geometric.data import Data

from rdkit import Chem, DataStructs
from rdkit.Chem import (
    AllChem,
    Descriptors,
    QED,
    rdFingerprintGenerator
)

# ------------------------------------------------------------------------------
# 1. Independent Diagnostics & Consistent Execution Anchors
# ------------------------------------------------------------------------------
warnings.filterwarnings("ignore", category=UserWarning)
warnings.filterwarnings("ignore", category=FutureWarning)

logging.basicConfig(
    level=logging.INFO,
    format="[%(asctime)s] [%(levelname)s] %(message)s",
    datefmt="%H:%M:%S"
)
logger = logging.getLogger("Cell2_DataIngestion")

DEMO_SEED = 42

# ------------------------------------------------------------------------------
# 2. External Database Backend Parameters (Storage Layer Only)
# ------------------------------------------------------------------------------
DB_CONFIG = {
    "host": "127.0.0.1",
    "port": 3306,
    "user": "root",
    "password": "",
    "database": "drugbank",
    "charset": "utf8mb4",
    "cursorclass": DictCursor
}

MAX_MOLECULES = 600                  # Working corpus demonstration subset
MAX_HEAVY_ATOMS = 100                # Small-molecule upper ceiling
MIN_HEAVY_ATOMS = 3                  # Exclude isolated counter-ions and single solvents
FP_DIM = 256                         # Auxiliary Morgan fingerprint dimension

# ------------------------------------------------------------------------------
# 3. Relational Query Execution (Extracting D_chem and D_val)
# ------------------------------------------------------------------------------
# In DrugBank SQL, kind is strictly 'SMILES'. We filter out long peptide chains
# via LENGTH(dp.value) < 250 to ensure small-molecule focus.
SMILES_QUERY = """
SELECT DISTINCT
    d.drug_pk,
    d.primary_drugbank_id AS drug_id,
    d.name AS drug_name,
    dp.value AS smiles
FROM drug AS d
INNER JOIN drug_property AS dp
    ON d.drug_pk = dp.drug_pk
WHERE dp.kind = 'SMILES'
  AND dp.value IS NOT NULL
  AND TRIM(dp.value) <> ''
  AND LENGTH(dp.value) < 250
LIMIT %s;
"""

TARGET_QUERY = """
SELECT DISTINCT
    d.drug_pk,
    d.primary_drugbank_id AS drug_id,
    d.name AS drug_name,
    dp.value AS smiles,
    i.interactant_pk,
    i.interactant_id,
    i.name AS target_name,
    i.organism AS target_organism,
    i.known_action,
    ip.polypeptide_id,
    ip.source AS polypeptide_source,
    p.amino_acid_sequence
FROM drug AS d
INNER JOIN drug_property AS dp
    ON d.drug_pk = dp.drug_pk
   AND dp.kind = 'SMILES'
INNER JOIN interactant AS i
    ON d.drug_pk = i.drug_pk
   AND LOWER(i.kind) = 'target'
INNER JOIN interactant_polypeptide AS ip
    ON i.interactant_pk = ip.interactant_pk
INNER JOIN polypeptide AS p
    ON ip.polypeptide_id = p.polypeptide_id
   AND ip.source = p.source
WHERE dp.value IS NOT NULL
  AND TRIM(dp.value) <> ''
  AND p.amino_acid_sequence IS NOT NULL
LIMIT %s;
"""

try:
    connection = pymysql.connect(**DB_CONFIG)
    logger.info(f"Connected to external storage backend: {DB_CONFIG['database']}")
    raw_chem_df = pd.read_sql(SMILES_QUERY, connection, params=(MAX_MOLECULES,))
    val_records_df = pd.read_sql(TARGET_QUERY, connection, params=(MAX_MOLECULES,))
    connection.close()
except Exception as exc:
    raise RuntimeError(
        f"\n[FATAL] Database connection or query failed on host {DB_CONFIG['host']}:{DB_CONFIG['port']}.\n"
        f"Verify that XAMPP MySQL is active and database '{DB_CONFIG['database']}' exists.\n"
        f"Original error: {exc}"
    )

if len(raw_chem_df) == 0:
    raise RuntimeError(
        "[FATAL] Zero chemical records retrieved from MySQL table 'drug_property' (kind='SMILES').\n"
        "Ensure the DrugBank database contains populated property records."
    )

logger.info(f"External Data Ingestion: {len(raw_chem_df)} raw chemical entries (D_chem), {len(val_records_df)} historical validation associations (D_val)")

# ------------------------------------------------------------------------------
# 4. Demonstration Preprocessing & Parsing
# ------------------------------------------------------------------------------
def sanitize_structure(smiles_str: str) -> str:
    """Optional preprocessing helper for demonstration runtime execution."""
    try:
        if not isinstance(smiles_str, str) or not smiles_str.strip():
            return None
        # Retain primary therapeutic fragment if salt counter-ions are present
        if "." in smiles_str:
            fragments = smiles_str.split(".")
            smiles_str = max(fragments, key=len)
        mol = Chem.MolFromSmiles(smiles_str)
        if mol is None:
            return None
        num_atoms = mol.GetNumHeavyAtoms()
        if num_atoms < MIN_HEAVY_ATOMS or (MAX_HEAVY_ATOMS is not None and num_atoms > MAX_HEAVY_ATOMS):
            return None
        return Chem.MolToSmiles(mol, canonical=True)
    except Exception:
        return None

raw_chem_df["canonical_smiles"] = raw_chem_df["smiles"].apply(sanitize_structure)
chem_corpus_df = raw_chem_df.dropna(subset=["canonical_smiles"]).drop_duplicates(subset=["canonical_smiles"]).reset_index(drop=True)

logger.info(f"Sanitized Small-Molecule Candidates (D_chem): {len(chem_corpus_df)}")

if len(chem_corpus_df) == 0:
    raise RuntimeError("[FATAL] All retrieved chemical structures were filtered out during sanitization.")

# ------------------------------------------------------------------------------
# 5. Atomic Feature Extraction (X) & Edge Attribute Extraction (E)
# ------------------------------------------------------------------------------
FP_GEN = rdFingerprintGenerator.GetMorganGenerator(radius=2, fpSize=FP_DIM)

def compute_atom_features(atom) -> list:
    """
    Extracts atomic feature vector X_i encoding atomic identity, hybridization,
    formal charge, and parity-sensitive chiral center indicators (FIG. 3).
    """
    hybridizations = [
        Chem.rdchem.HybridizationType.SP,
        Chem.rdchem.HybridizationType.SP2,
        Chem.rdchem.HybridizationType.SP3
    ]
    hyb_one_hot = [float(atom.GetHybridization() == h) for h in hybridizations]
    hyb_one_hot.append(float(atom.GetHybridization() not in hybridizations))

    chiral_tags = [
        Chem.rdchem.ChiralType.CHI_UNSPECIFIED,
        Chem.rdchem.ChiralType.CHI_TETRAHEDRAL_CW,
        Chem.rdchem.ChiralType.CHI_TETRAHEDRAL_CCW
    ]
    chiral_one_hot = [float(atom.GetChiralTag() == c) for c in chiral_tags]
    chiral_one_hot.append(float(atom.GetChiralTag() not in chiral_tags))

    return [
        float(atom.GetAtomicNum()) / 100.0,
        float(atom.GetFormalCharge()) / 5.0,
        float(atom.GetTotalDegree()) / 6.0,
        float(atom.GetIsAromatic()),
        *hyb_one_hot,
        *chiral_one_hot
    ]

def compute_bond_features(bond) -> list:
    """
    Extracts edge attribute vector e_ij encoding bond type, conjugation status,
    and stereochemical configuration flags (Patent FIG. 3 / Claim 2).
    """
    b_type = bond.GetBondType()
    stereo = bond.GetStereo()
    return [
        float(b_type == Chem.rdchem.BondType.SINGLE),
        float(b_type == Chem.rdchem.BondType.DOUBLE),
        float(b_type == Chem.rdchem.BondType.TRIPLE),
        float(b_type == Chem.rdchem.BondType.AROMATIC),
        float(bond.GetIsConjugated()),
        float(stereo == Chem.rdchem.BondStereo.STEREONONE),
        float(stereo in [Chem.rdchem.BondStereo.STEREOZ, Chem.rdchem.BondStereo.STEREOCIS]),
        float(stereo in [Chem.rdchem.BondStereo.STEREOE, Chem.rdchem.BondStereo.STEREOTRANS])
    ]

# ------------------------------------------------------------------------------
# 6. Attributed Spatial Graph Builder G = (V, E, X, R)
# ------------------------------------------------------------------------------
def smiles_to_spatial_graph(smiles: str, drug_id: str = "UNKNOWN", drug_name: str = "UNKNOWN") -> Data:
    """
    Constructs an attributed 3D molecular graph G = (V, E, X, R) conforming to FIG. 3.
    Uses RDKit ETKDGv3 distance geometry with a planar 2D-to-3D projection fallback.
    """
    try:
        mol = Chem.MolFromSmiles(smiles)
        if mol is None or mol.GetNumHeavyAtoms() == 0:
            return None

        coords = None

        # Embodiment 1: 3D Conformer via ETKDGv3
        try:
            mol_h = Chem.AddHs(mol)
            params = AllChem.ETKDGv3()
            params.randomSeed = DEMO_SEED
            params.maxAttempts = 15
            params.useRandomCoords = True

            if AllChem.EmbedMolecule(mol_h, params) == 0:
                try:
                    if AllChem.MMFFHasAllMoleculeParams(mol_h):
                        AllChem.MMFFOptimizeMolecule(mol_h, maxIters=30)
                except Exception:
                    pass
                mol_no_h = Chem.RemoveHs(mol_h)
                if mol_no_h.GetNumConformers() > 0:
                    conf = mol_no_h.GetConformer()
                    coords = [list(conf.GetAtomPosition(i)) for i in range(mol_no_h.GetNumAtoms())]
                    mol = mol_no_h
        except Exception:
            coords = None

        # Embodiment 2: Planar 2D layout projected into 3D space (Z = 0.0) fallback
        if coords is None:
            AllChem.Compute2DCoords(mol)
            if mol.GetNumConformers() > 0:
                conf = mol.GetConformer()
                coords = [[conf.GetAtomPosition(i).x, conf.GetAtomPosition(i).y, 0.0] for i in range(mol.GetNumAtoms())]
            else:
                coords = [[float(i), 0.0, 0.0] for i in range(mol.GetNumAtoms())]

        num_atoms = mol.GetNumAtoms()

        # Patent Entity: Node Feature Matrix X in R^(N x F_v)
        x_features = [compute_atom_features(mol.GetAtomWithIdx(i)) for i in range(num_atoms)]
        x = torch.tensor(x_features, dtype=torch.float32)

        # Patent Entity: Cartesian Spatial Coordinate Matrix R in R^(N x 3)
        pos = torch.tensor(coords, dtype=torch.float32)
        pos = pos - pos.mean(dim=0, keepdim=True)  # Coordinate origin centering

        # Patent Entity: Edge Index E and Attribute Tensor e_ij in R^F_e
        src_nodes, dst_nodes, edge_attributes = [], [], []
        for bond in mol.GetBonds():
            u = bond.GetBeginAtomIdx()
            v = bond.GetEndAtomIdx()
            attr = compute_bond_features(bond)
            src_nodes.extend([u, v])
            dst_nodes.extend([v, u])
            edge_attributes.extend([attr, attr])

        if len(src_nodes) == 0:
            edge_index = torch.empty((2, 0), dtype=torch.long)
            edge_attr = torch.empty((0, 8), dtype=torch.float32)
        else:
            edge_index = torch.tensor([src_nodes, dst_nodes], dtype=torch.long)
            edge_attr = torch.tensor(edge_attributes, dtype=torch.float32)

        # Auxiliary Morgan fingerprint vector [256]
        fp = FP_GEN.GetFingerprint(mol)
        fp_arr = np.zeros((FP_DIM,), dtype=np.float32)
        DataStructs.ConvertToNumpyArray(fp, fp_arr)
        fp_tensor = torch.tensor(fp_arr, dtype=torch.float32)

        # Attributed Spatial Graph Instance G = (V, E, X, R)
        graph = Data(
            x=x,
            pos=pos,
            edge_index=edge_index,
            edge_attr=edge_attr
        )
        graph.fp = fp_tensor
        graph.smiles = Chem.MolToSmiles(mol, canonical=True)
        graph.drug_id = str(drug_id)
        graph.drug_name = str(drug_name)

        return graph

    except Exception:
        return None

# ------------------------------------------------------------------------------
# 7. Attributed Graph Corpus Construction (D_chem)
# ------------------------------------------------------------------------------
graph_corpus_chem = []
for record in tqdm(chem_corpus_df.itertuples(index=False), total=len(chem_corpus_df), desc="Compiling 3D Spatial Graphs (D_chem)"):
    g = smiles_to_spatial_graph(
        smiles=record.canonical_smiles,
        drug_id=record.drug_id,
        drug_name=record.drug_name
    )
    if g is not None:
        graph_corpus_chem.append(g)

if len(graph_corpus_chem) == 0:
    raise RuntimeError("[FATAL] Spatial graph compilation yielded 0 valid graphs. Inspect chemical structures.")

NODE_FEAT_DIM = graph_corpus_chem[0].x.shape[1]
EDGE_FEAT_DIM = graph_corpus_chem[0].edge_attr.shape[1]

# ------------------------------------------------------------------------------
# 8. Baseline Partitioning for Closed-Loop Continual Learning
# ------------------------------------------------------------------------------
random.seed(DEMO_SEED)
shuffled_corpus = list(graph_corpus_chem)
random.shuffle(shuffled_corpus)

split_idx = int(0.85 * len(shuffled_corpus))
train_graphs = shuffled_corpus[:split_idx]
validation_graphs = shuffled_corpus[split_idx:]

# ------------------------------------------------------------------------------
# 9. Summary Telemetry
# ------------------------------------------------------------------------------
print("=" * 80)
print("CELL 2 EXECUTION COMPLETE: 3D ATTRIBUTED GRAPH CORPUS CONSTRUCTED")
print("=" * 80)
print(f"Total Spatial Graphs Compiled (|D_chem|) : {len(graph_corpus_chem)} spatial graphs G = (V, E, X, R)")
print(f"  • Baseline Pretraining Subset (Train)   : {len(train_graphs)}")
print(f"  • Historical Evaluation Subset (Val)    : {len(validation_graphs)}")
print(f"Historical Validation Records (|D_val|)   : {len(val_records_df)} drug-target associations")
print(f"Atomic Feature Matrix Dimension (F_v)     : {NODE_FEAT_DIM} features per node")
print(f"Edge Attribute Tensor Dimension (F_e)     : {EDGE_FEAT_DIM} features per edge")
print(f"Cartesian Coordinate Matrix Dimension (R) : R^(N x 3) [Origin Centered]")
print("-" * 80)
sample_g = graph_corpus_chem[0]
print(f"Representative Graph Sample [{sample_g.drug_id} - {sample_g.drug_name}]:")
print(f"  • Node tensor X shape       : {list(sample_g.x.shape)}")
print(f"  • Coordinate tensor R shape : {list(sample_g.pos.shape)}")
print(f"  • Edge index E shape        : {list(sample_g.edge_index.shape)}")
print(f"  • Edge attribute tensor     : {list(sample_g.edge_attr.shape)}")
print(f"  • Morgan fingerprint tensor : {list(sample_g.fp.shape)}")
print("=" * 80)
```

# Cell 2: DrugBank Database Ingestion & 3D Molecular Graph Preparation

## Objective

- Required tables:
  - `drug`
  - `drug_property`

## Processing Steps
1. Connect to the DrugBank MySQL database using PyMySQL.
2. Retrieve:
   - Chemical structures (SMILES) for the baseline chemical corpus (**D<sub>chem</sub>**).
   - Drug–target associations and protein sequences for validation records (**D<sub>val</sub>**).

   - Remove invalid entries.
   - Select major molecular fragments.

\[
G = (V,E,X,R)
\]

where:
- **V** = molecular atoms  
- **E** = chemical bonds  
- **X** = atom-level features  
- **R** = 3D Cartesian coordinates generated using RDKit ETKDGv3.

5. Store graph objects in PyTorch Geometric format for molecular encoder training.

## Output

# Cell 3: DrugBank Dataset Extraction

## Objective
Extract molecular chemical structures and historical drug–target interaction records from the DrugBank relational database storage layer to construct the baseline chemical structure corpus ($\mathcal{D}_{\text{chem}}$) and historical validation dataset ($\mathcal{D}_{\text{val}}$).

## Patent Module Mapping
- **System 100:** Closed-Loop Molecular Optimization System
- **Corpus Preparation:**
  - $\mathcal{D}_{\text{chem}}$: Baseline chemical structure corpus for unsupervised structural prior loss ($\mathcal{L}_{\text{NELBO}}$) and structural Fisher Information anchoring ($F_j$) in Module 108.
  - $\mathcal{D}_{\text{val}}$: Historical validation records for multi-objective bioactivity adaptation ($\mathcal{L}_{\text{task}}$) and predictive memory preservation in Module 108.

## Input Data
- Active MySQL database connection handle (`connection`) established in Cell 2.
- Demonstration extraction ceiling parameter (`MAX_MOLECULES`).

## Processing Steps
1. Query genuine small-molecule chemical structures ($\mathcal{D}_{\text{chem}}$) from `drug` and `drug_property` where `kind = 'SMILES'`.
2. Query historical drug–target bioactivity associations ($\mathcal{D}_{\text{val}}$) by joining `drug`, `drug_property`, `interactant` (`kind = 'target'`), `interactant_polypeptide`, and `polypeptide` on the composite key:
   $$(\text{polypeptide\_id}, \text{source})$$
3. Convert SQL query outputs directly into Pandas DataFrames (`chem_df`, `target_df`) using `pd.read_sql`.
4. Perform preliminary demonstration cleaning:
   - Filter empty or non-chemical strings.
   - Retain parent active scaffolds from salt mixtures (`.`).
   - Remove duplicate structures.
5. Close the database connection handle (`connection.close()`) after data extraction completes.

## Mathematical Operation
Relational algebraic projection, selection, and composite inner-join filtering:
$$\mathcal{D}_{\text{chem}} = \pi_{\text{drug\_id, smiles}} \left( \sigma_{\text{kind}=\text{'SMILES'}} (\text{drug} \bowtie \text{drug\_property}) \right)$$
$$\mathcal{D}_{\text{val}} = \pi_{\text{drug\_id, target\_name, sequence}} \left( \sigma_{\text{kind}=\text{'target'}} (\text{drug} \bowtie \text{drug\_property} \bowtie \text{interactant} \bowtie \text{interactant\_polypeptide} \bowtie \text{polypeptide}) \right)$$

## Expected Output
Pandas DataFrames held in memory:
- `chem_df`: Extracted and sanitized chemical structure records ($\mathcal{D}_{\text{chem}}$).
- `target_df`: Validated drug–target and protein sequence interaction records ($\mathcal{D}_{\text{val}}$).

## Saved Results
In-memory objects `chem_df` and `target_df` ready for 3D conformer embedding and spatial graph compilation $G = (V, E, X, R)$ in Cell 4.

```python
# ==============================================================================
# NOTEBOOK CELL 3: DATASET EXTRACTION (D_chem AND D_val)
# ==============================================================================
# Objective:
# Extract chemical structural definitions and historical target interaction records
# from the relational storage backend to construct the baseline chemical corpus
# (D_chem) and historical bioactivity dataset (D_val).
#
# Patent Module Mapping:
# - Input Preparation for System 100
# - Corpus preparation:
#     * Baseline chemical structure corpus (D_chem) for Module 104 & Module 108
#     * Historical validation dataset (D_val) for Engine 106 & Module 108
#
# Input Data:
# Active database connection handle `connection` from Cell 2.
# Demonstration extraction limit (`MAX_MOLECULES`).
# ==============================================================================

import logging
import warnings
import pymysql
import pandas as pd
from rdkit import Chem

# ------------------------------------------------------------------------------
# 1. Diagnostics Setup & Parameter Verification
# ------------------------------------------------------------------------------
warnings.filterwarnings("ignore", category=UserWarning)
warnings.filterwarnings("ignore", category=FutureWarning)

logging.basicConfig(
    level=logging.INFO,
    format="[%(asctime)s] [%(levelname)s] %(message)s",
    datefmt="%H:%M:%S"
)
logger = logging.getLogger("Cell3_DatasetExtraction")

MAX_MOLECULES = 800
MIN_HEAVY_ATOMS = 4
MAX_HEAVY_ATOMS = 100

# ------------------------------------------------------------------------------
# 2. Relational Query Definitions (D_chem and D_val)
# ------------------------------------------------------------------------------
# Baseline Chemical Structure Corpus (D_chem):
# Queries dp.kind = 'SMILES' directly without restrictive property_type filter
SMILES_QUERY = """
SELECT DISTINCT
    d.drug_pk,
    d.primary_drugbank_id AS drug_id,
    d.name AS drug_name,
    dp.value AS smiles
FROM drug AS d
INNER JOIN drug_property AS dp
    ON d.drug_pk = dp.drug_pk
WHERE dp.kind = 'SMILES'
  AND dp.value IS NOT NULL
  AND TRIM(dp.value) <> ''
  AND LENGTH(dp.value) BETWEEN 10 AND 200
LIMIT %s;
"""

# Historical Bioactivity Validation Dataset (D_val):
# Composite key join on (polypeptide_id, source) as defined in the DrugBank schema
TARGET_QUERY = """
SELECT DISTINCT
    d.drug_pk,
    d.primary_drugbank_id AS drug_id,
    d.name AS drug_name,
    dp.value AS smiles,
    i.interactant_pk,
    i.interactant_id,
    i.name AS target_name,
    i.organism AS target_organism,
    i.known_action,
    ip.polypeptide_id,
    ip.source AS polypeptide_source,
    p.amino_acid_sequence
FROM drug AS d
INNER JOIN drug_property AS dp
    ON d.drug_pk = dp.drug_pk
   AND dp.kind = 'SMILES'
INNER JOIN interactant AS i
    ON d.drug_pk = i.drug_pk
   AND LOWER(i.kind) = 'target'
INNER JOIN interactant_polypeptide AS ip
    ON i.interactant_pk = ip.interactant_pk
INNER JOIN polypeptide AS p
    ON ip.polypeptide_id = p.polypeptide_id
   AND ip.source = p.source
WHERE dp.value IS NOT NULL
  AND TRIM(dp.value) <> ''
  AND LENGTH(dp.value) BETWEEN 10 AND 200
  AND p.amino_acid_sequence IS NOT NULL
LIMIT %s;
"""

# ------------------------------------------------------------------------------
# 3. Query Execution on Active Connection & Graceful Disconnect
# ------------------------------------------------------------------------------
if "connection" not in globals() or not connection.open:
    connection = pymysql.connect(**DB_CONFIG)
    logger.info("Re-established socket connection to MySQL database.")

try:
    logger.info("Extracting baseline chemical structure corpus (D_chem)...")
    chem_df = pd.read_sql(SMILES_QUERY, connection, params=(MAX_MOLECULES,))
    
    logger.info("Extracting historical drug-target validation records (D_val)...")
    target_df = pd.read_sql(TARGET_QUERY, connection, params=(MAX_MOLECULES,))

finally:
    connection.close()
    logger.info("Database socket connection safely closed.")

logger.info(f"Raw chemical records ingested (D_chem): {len(chem_df)}")
logger.info(f"Drug-target pairs ingested (D_val): {len(target_df)}")

# ------------------------------------------------------------------------------
# 4. SMILES Sanitization (Parent Scaffold Isolation)
# ------------------------------------------------------------------------------
def sanitize_smiles(raw_smiles: str) -> str:
    """Retains the primary therapeutic scaffold from salts and canonicalizes notation."""
    try:
        if not isinstance(raw_smiles, str) or not raw_smiles.strip():
            return None
        s = raw_smiles.strip()
        # Retain largest fragment if multi-component mixture/salt exists
        if "." in s:
            fragments = s.split(".")
            s = max(fragments, key=len)
        mol = Chem.MolFromSmiles(s)
        if mol is None:
            return None
        num_atoms = mol.GetNumHeavyAtoms()
        if num_atoms < MIN_HEAVY_ATOMS or (MAX_HEAVY_ATOMS is not None and num_atoms > MAX_HEAVY_ATOMS):
            return None
        return Chem.MolToSmiles(mol, canonical=True)
    except Exception:
        return None

chem_df["canonical_smiles"] = chem_df["smiles"].apply(sanitize_smiles)
chem_df = chem_df.dropna(subset=["canonical_smiles"]).drop_duplicates(subset=["canonical_smiles"]).reset_index(drop=True)

logger.info(f"Sanitized small-molecule drug candidates (D_chem): {len(chem_df)}")

# ------------------------------------------------------------------------------
# 5. Summary Telemetry
# ------------------------------------------------------------------------------
print("=" * 80)
print("CELL 3 EXECUTION COMPLETE: DATASETS EXTRACTED AND SANITIZED")
print("=" * 80)
print(f"Baseline Chemical Corpus (|D_chem|) : {len(chem_df)} sanitized small molecules")
print(f"Historical Validation Records (|D_val|): {len(target_df)} drug-target associations")
print("-" * 80)
print("Baseline Chemical Corpus Preview (chem_df):")
display(chem_df[["drug_id", "drug_name", "canonical_smiles"]].head(3))

if len(target_df) > 0:
    print("\nHistorical Validation Dataset Preview (target_df):")
    display(target_df[["drug_id", "drug_name", "interactant_id", "target_name", "polypeptide_id"]].head(3))
print("=" * 80)
```
